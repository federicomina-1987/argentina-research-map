## Open file, normalize values, extract filiation and group by author

```python
df = pd.read_csv(r'C:\Users\Usuario\Desktop\conicet_emb_umap_2026-09-02.csv')
df
```

```python
# erase the next line
df.drop(columns=['x', 'y'])


#prepare embeddings columns
def parse_embedding_string(emb_str):
    if isinstance(emb_str, str):
        # Remove the brackets and newlines, then split by spaces into floats
        clean_str = emb_str.replace('[', '').replace(']', '').replace('\n', ' ')
        return np.fromstring(clean_str, sep=' ')
    return emb_str # Returns as-is if it's already an array

# 2. Apply the fix to the dataframe
df['emb'] = df['emb'].apply(parse_embedding_string)

# 3. Now build the matrix as normal

embeddings_matrix = np.vstack(df['emb'].values)
```

```python
# 1) Explode the column based on the '|' separator
df['filiation'] = df['filiation'].astype(str).str.split('|')
df = df.explode('filiation')

# 2) Clean starting and trailing spaces
df['filiation'] = df['filiation'].str.strip()

# 3) Eliminate non-Argentina affiliations
len_before_filter = len(df)
# The regex (?i) ignores case, \.? allows an optional dot, and $ ensures it's at the end
mask_argentina = df['filiation'].str.contains(r'(?i)argentina\.?$', na=False)
df = df[mask_argentina]
len_after_filter = len(df)

print(f"Rows before Argentina filter: {len_before_filter}")
print(f"Rows after Argentina filter:  {len_after_filter}")

# 4) Extract the name
# Captures everything after 'Fil:' or 'Fil.:' until the first dot
df['name'] = df['filiation'].str.extract(r'^Fil\.?:\s*([^.]+)')
df['name'] = df['name'].str.strip()

# 5) Extract 'lugar_trabajo'
# Captures everything from the dot after the name until the semi-colon
df['lugar_trabajo'] = df['filiation'].str.extract(r'^Fil\.?:\s*[^.]+\.\s*(.*?)\s*;')
df['lugar_trabajo'] = df['lugar_trabajo'].str.strip()

df.shape
```

```python
# Clean up any rows where extraction failed
df = df.dropna(subset=['name'])

# 6) Group values by normalized name but preserve the accented version
len_before_grouping = len(df)

def normalize_name(s):
    """Converts to lowercase and strips accents (e.g., 'í' -> 'i') for grouping"""
    if pd.isna(s): return s
    s = str(s).lower()
    return ''.join(c for c in unicodedata.normalize('NFD', s) if unicodedata.category(c) != 'Mn')

df['name_norm'] = df['name'].apply(normalize_name)

# Sort by the number of accented characters descending. 
# This guarantees that when we use .first() in the groupby, the version WITH accents is kept.
df['accent_count'] = df['name'].str.count(r'[áéíóúÁÉÍÓÚñÑ]')
df = df.sort_values(by=['name_norm', 'accent_count'], ascending=[True, False])

# 7) Group by author and apply mathematical Mean Pooling to the embeddings
grouped_df = df.groupby('name_norm').agg(
    name=('name', 'first'),                   # Keeps the highest-accented version of the name
    lugar_trabajo=('lugar_trabajo', 'first'), # Keeps their primary/first listed workplace
    paper_count=('name', 'count'),            # Tracks how many papers the author has
    text=('text', ' '.join),                  # Concatenates all texts into a single string separated by space

    # THE MATH: Stacks all 768-D arrays for this author into a matrix and calculates the vertical mean.
    # This creates the exact semantic centroid (average) of their entire publishing history.
    emb=('emb', lambda vectors: np.mean(np.vstack(vectors), axis=0))
).reset_index(drop=True)

len_after_grouping = len(grouped_df)

print(f"Rows before grouping by author: {len_before_grouping}")
print(f"Rows after grouping by author:  {len_after_grouping}")
```

```python
# eliminate authors with only one paper ot get a cleaner visualization

grouped_df = grouped_df[grouped_df['paper_count'] > 1]
grouped_df.shape
```

# Calculate TD-IDF

```python
# Download the stopwords dataset (you only need to run this once)
nltk.download('stopwords')

# Load and combine English and Spanish stopwords
english_stops = stopwords.words('english')
spanish_stops = stopwords.words('spanish')

# Combine and remove any potential duplicates by converting to a set, then back to a list
combined_stopwords = list(set(english_stops + spanish_stops))

# Initialize TF-IDF Vectorizer with the combined list
vectorizer = TfidfVectorizer(
    stop_words=combined_stopwords, 
    max_df=0.85, # Ignores words appearing in >85% of authors
    min_df=1     
)

# Fit and transform the concatenated text
tfidf_matrix = vectorizer.fit_transform(grouped_df['text'])
feature_names = np.array(vectorizer.get_feature_names_out())

# Define a function to extract top words with a threshold
def get_top_words(row_idx, matrix, features, top_n=7, threshold=0.10):
    """Extracts top N words for a specific row, filtering out scores below the threshold."""
    row = matrix.getrow(row_idx)
    
    # row.indices contains the vocabulary indices of non-zero TF-IDF scores
    # row.data contains the actual TF-IDF scores
    indices = row.indices
    scores = row.data
    
    # Apply the significance threshold
    valid_mask = scores >= threshold
    filtered_indices = indices[valid_mask]
    filtered_scores = scores[valid_mask]
    
    if len(filtered_scores) == 0:
        return ""
    
    # Sort by score in descending order and slice the top_n
    sorted_order = np.argsort(filtered_scores)[::-1]
    top_indices = filtered_indices[sorted_order][:top_n]
    
    # Map back to strings and join with commas
    return ", ".join(features[top_indices])

# Apply to the dataframe
grouped_df['top_keywords'] = [
    get_top_words(i, tfidf_matrix, feature_names, top_n=7, threshold=0.08) 
    for i in range(grouped_df.shape[0])
]
```

# Calculate UMAP coordinates to convert embeddings to 2d positions

```python
try:
    grouped_df.drop(columns=['x', 'y'])
except:
    pass


embeddings_matrix = np.vstack(grouped_df['emb'].values)

# Add microscopic noise to break exact duplicates
print("Adding microscopic noise to prevent UMAP crashes...")
noise = np.random.normal(0, 1e-7, embeddings_matrix.shape)
embeddings_clean = embeddings_matrix + noise

```

```python


print("Calculating 2D UMAP (Macro Global Separation)...")
reducer = umap.UMAP(
    n_components=2, 
    metric='cosine', 
    n_neighbors=60,   # High value forces global structure preservation over local overlapping
    min_dist=0.1,      # Slightly higher to prevent clusters from collapsing into singular points
    n_jobs=-1,
    densmap=False      # DISABLED
)

# Fit and transform
umap_2d = reducer.fit_transform(embeddings_clean)

grouped_df['x'] = umap_2d[:, 0]
grouped_df['y'] = umap_2d[:, 1]
```

```python
# Manually erase outliers

# Define threshold limits based on your desired display window
x_min, x_max = -8.1, 7.0
y_min, y_max = -3.5, 9.5

# Identify any point outside the valid ranges
outlier_mask = (
    (grouped_df['x'] < x_min) | (grouped_df['x'] > x_max) |
    (grouped_df['y'] < y_min) | (grouped_df['y'] > y_max)
)

# Remove all outliers
grouped_df = grouped_df[~outlier_mask].copy().reset_index(drop=True)

print(f"Removed {outlier_mask.sum()} points falling outside X:[{x_min}, {x_max}] or Y:[{y_min}, {y_max}].")
```

## Identify clusters

```python
from collections import Counter
from IPython.display import display

def get_top_cluster_words(keywords_series, top_n=30):
    """Combines all author keywords in a cluster and counts the most frequent."""
    all_text = ", ".join(keywords_series.dropna().astype(str))
    words = [word.strip() for word in all_text.split(',') if word.strip()]
    
    word_counts = Counter(words)
    top_words = [word for word, count in word_counts.most_common(top_n)]
    
    return ", ".join(top_words)

# Group by cluster and apply the counting function
cluster_summary = clean_df.groupby('cluster')['top_keywords'].apply(
    lambda x: get_top_cluster_words(x, top_n=30)
).reset_index()

cluster_summary.rename(columns={'top_keywords': 'top_30_words'}, inplace=True)

# Temporarily remove display limits just for this block
with pd.option_context('display.max_colwidth', None, 'display.max_rows', None):
    display(cluster_summary)
```
